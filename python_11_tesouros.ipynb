#Tesouros

```python
tesouros = int(input('Informe o número de tesouros: '))
membros = int(input('Informe o número de membros: '))
bonus = 10
quantidade = (tesouros - bonus) // membros
lider = bonus + quantidade
print(f'Lider: {lider}')
print(f'Quantidade os recebem {quantidade}')
restante = tesouros % membros
guia = quantidade + restante
print(f'Guia recebe {guia}')
```
