##Um jardim tem o formato de um retângulo com 12 metros de comprimento e 7 metros de largura.
Qual é o perímetro do jardim?

Fórmula: Perímetro = 2 × (comprimento + largura)

```python
comprimento = 12
largura = 7

perimetro = 2 * (comprimento + largura)

print("O perímetro do jardim é:", perimetro, "metros")
```
