Uma escola está promovendo uma campanha de incentivo à leitura em que cada estudante recebe pontos ao concluir livros. Esses pontos podem ser trocados por diferentes prêmios da escola, e as categorias de livros têm valores de pontos diferentes. Livros de ficção valem 8 pontos, livros de não ficção valem 8 pontos, e livros infantis valem 6 pontos.
O problema é que Rodrigo leu um livro de cada categoria e agora quer saber quantos pontos acumulou com sua leitura.

```python
## PONTOS RODRIGO

livro_ficcao = 10
livro_nficcao = 8
livro_infantil = 6
pontos_rodrigo = livro_ficcao + livro_nficcao + livro_infantil
print(f"Os pontos de Rodrigo são: {pontos_rodrigo}")

```

Ana também participou da campanha e leu dois livros de ficção e cinco livros infantis. Como podemos calcular o total de pontos de Ana?


```python
## PONTOS ANA

livro_ficcao = 10
livro_nficcao = 8
livro_infantil = 6
pontos_rodrigo = livro_ficcao + livro_nficcao + livro_infantil
pontos_ana = (livro_ficcao * 2) + (livro_infantil * 5)
pontos_total = pontos_rodrigo + pontos_ana

print(f"Os pontos de Rodrigo são: {pontos_rodrigo}")

print(f"Os pontos totais de Ana são: {pontos_ana}")
```

 É possível calcular o total de pontos acumulados por Rodrigo e Ana juntos? Este é um desafio interessante, que pode exigir o uso da soma mais de uma vez, ou até mesmo multiplicação.

```python
## PONTOS ACUMULADOS RODRIGO E ANA

livro_ficcao = 10
livro_nficcao = 8
livro_infantil = 6
pontos_rodrigo = livro_ficcao + livro_nficcao + livro_infantil
pontos_ana = (livro_ficcao * 2) + (livro_infantil * 5)
pontos_total = pontos_rodrigo + pontos_ana

print(f"Os pontos de Rodrigo são: {pontos_rodrigo}")

print(f"Os pontos totais de Ana são: {pontos_ana}")

print(f"O total de pontos acumulados de Ana e Rodrigo é: {pontos_total}")
```
