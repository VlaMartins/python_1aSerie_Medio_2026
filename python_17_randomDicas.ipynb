```python
import random

# Corrigido para 'dicas' em minúsculo para padronizar
dicas = [
    "Pratique atividade física.",
    "Beba água.",
    "Durma ao menos 8h por noite.",
    "Evite telas antes de ir deitar-se."
]

while True:
    # Adicionado um espaço no input para melhor formatação visual
    pedido = input("Digite 'dica' para receber uma dica ou 'sair' para encerrar: ").lower()

    if pedido == "dica":
        dica = random.choice(dicas)
        print(f"A dica é: {dica}\n") # \n adiciona uma linha em branco para organizar o terminal

    elif pedido == "sair":
        print("Programa encerrado. Até mais!")
        break # Interrompe o loop while e fecha o programa

    else:
        print("Não conheço esse comando. Tente de novo!\n")
```
