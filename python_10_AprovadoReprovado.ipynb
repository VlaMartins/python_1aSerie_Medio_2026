#Aula 3 - Aprovado / Reprovado

Escreva um programa onde o estudante informe sua nota. Se a nota for maior ou igual a 6.0 passe uma mensagem dizendo que ele foi aprovado, senão avise que está reprovado.

```python
nota = float(input("Digite sua nota: "))

if nota >= 7.0: print("Aprovado")
else: print("Reprovado")
```
